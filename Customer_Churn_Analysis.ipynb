```python
result_payment_check = pd.read_sql(
    """
    SELECT
        CASE
            WHEN DBMS_LOB.SUBSTR("Payment_Method", 100, 1) IS NULL
              OR DBMS_LOB.SUBSTR("Payment_Method", 100, 1) = 'Unknown'
            THEN 'Unknown/Missing'
            ELSE DBMS_LOB.SUBSTR("Payment_Method", 100, 1)
        END AS PAYMENT_METHOD,
        COUNT(*) AS TOTAL_CUSTOMERS,
        SUM(
            CASE
                WHEN DBMS_LOB.SUBSTR("Churn", 100, 1) = 'Yes'
                THEN 1 ELSE 0
            END
        ) AS CHURNED_CUSTOMERS,
        ROUND(
            100 * SUM(
                CASE
                    WHEN DBMS_LOB.SUBSTR("Churn", 100, 1) = 'Yes'
                    THEN 1 ELSE 0
                END
            ) / COUNT(*),
            2
        ) AS CHURN_RATE_PERCENT
    FROM CUSTOMER_CHURN
    WHERE DBMS_LOB.SUBSTR("Churn", 100, 1) IS NOT NULL
    GROUP BY
        CASE
            WHEN DBMS_LOB.SUBSTR("Payment_Method", 100, 1) IS NULL
              OR DBMS_LOB.SUBSTR("Payment_Method", 100, 1) = 'Unknown'
            THEN 'Unknown/Missing'
            ELSE DBMS_LOB.SUBSTR("Payment_Method", 100, 1)
        END
    ORDER BY PAYMENT_METHOD
    """,
    engine
)

display(result_payment_check)
```
